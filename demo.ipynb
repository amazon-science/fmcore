```python
class AddLol(Cleaner):
    class Params(Cleaner.Params):
        reps: int = 1
        col: str = 'generations'
    
    def clean(
            self,
            data: pd.DataFrame,
    ) -> pd.DataFrame:
        for _ in range(self.params.reps):
            data[self.params.col] = data[self.params.col].astype(str) + ' lol'
        return data
```

```python
@actor
class CounterActor:
    def __init__(self, start_value=0):
        self.count = start_value

    def increment(self, value=1):
        self.count += value
        return self.count

    def get_count(self):
        return self.count

    def reset(self):
        self.count = 0
        return self.count

    def count_to(self, i):
        count = self.count
        for _ in range(int(i)):
            count += 1
        return count
```

```python
from concurrent.futures import Executor, Future, ThreadPoolExecutor
from joblib import Parallel, delayed

class JoblibExecutor(Executor):
    def __init__(self, max_workers=None, backend='loky', verbose=0):
        """
        A simple executor-like class that uses joblib.Parallel under the hood.
        
        Parameters
        ----------
        max_workers : int, optional
            The number of jobs to run in parallel (default: all CPUs).
        backend : str, optional
            The parallelization backend to use (e.g. 'loky', 'multiprocessing', 'threading').
            If None, joblib chooses the best available default.
        verbose : int, optional
            The verbosity level for joblib.Parallel.
        """
        self.max_workers = max_workers
        self.backend = backend
        self.verbose = verbose
        self._shutdown = False
        # Use a thread pool to handle running parallel tasks asynchronously
        self._thread_pool = ThreadPoolExecutor(max_workers=max_workers+1)

    def submit(self, fn, *args, **kwargs):
        """
        Schedules the fn(*args, **kwargs) to be executed and returns a Future.
        """
        if self._shutdown:
            raise RuntimeError("Cannot schedule new futures after shutdown.")

        future = Future()

        def run_task():
            try:
                # We run a single task via Parallel, which returns a list of results.
                # Since we're only giving it one function call, we get a single-result list.
                results = Parallel(n_jobs=self.max_workers, backend=self.backend, verbose=self.verbose)(
                    [delayed(fn)(*args, **kwargs)]
                )
                future.set_result(results[0])
            except Exception as e:
                future.set_exception(e)

        # Submit the parallel run to the thread pool so that it doesn't block
        self._thread_pool.submit(run_task)
        return future

    def shutdown(self, wait=True):
        """
        Clean up the executor. If wait=True, wait for all pending futures to finish.
        """
        self._shutdown = True
        self._thread_pool.shutdown(wait=wait)

    def map(self, fn, *iterables, timeout=None, chunksize=1):
        """
        An optional map method, similar to ProcessPoolExecutor.map, but runs all at once.
        
        Note: This method is synchronous. It runs the entire mapping at once using Parallel.
        """
        if self._shutdown:
            raise RuntimeError("Cannot schedule new futures after shutdown.")

        # Prepare the tasks
        # joblib Parallel likes a list of delayed calls
        tasks = (delayed(fn)(*args) for args in zip(*iterables))
        results = Parallel(n_jobs=self.max_workers, backend=self.backend, verbose=self.verbose)(tasks)
        return results
```
